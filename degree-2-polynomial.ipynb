```python

# Degree 2 Polynomial Regression

# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Step 2: Create Dataset
data = {
    'X': [-2, -1, 0, 1, 2],
    'Y': [9, 2, 1, 6, 17]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Step 3: Separate Input and Output
X = df[['X']]
y = df['Y']

# Step 4: Generate Degree 2 Polynomial Features
poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

print("\nPolynomial Features:")
print(X_poly)

# Step 5: Train Polynomial Regression Model
model = LinearRegression()
model.fit(X_poly, y)

# Step 6: Predict Values
y_pred = model.predict(X_poly)

print("\nActual Values:")
print(list(y))

print("\nPredicted Values:")
print(np.round(y_pred, 2))

# Step 7: Evaluate Model
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("\nMean Squared Error:", round(mse, 4))
print("R2 Score:", round(r2, 4))

# Step 8: Display Polynomial Equation
coefficients = model.coef_
intercept = model.intercept_

print("\nDegree 2 Polynomial Equation:")
print(f"Y = {intercept:.2f} + {coefficients[1]:.2f}X + {coefficients[2]:.2f}X^2")

# Step 9: Visualize Results
X_range = np.linspace(
    X['X'].min(), X['X'].max(), 200
).reshape(-1, 1)

X_range_poly = poly.transform(X_range)
y_range_pred = model.predict(X_range_poly)

plt.figure(figsize=(8, 5))

plt.scatter(X, y, color='blue', label='Actual Data')

plt.plot(
    X_range, y_range_pred,
    color='red',
    label='Degree 2 Polynomial Regression'
)

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Degree 2 Polynomial Regression')
plt.legend()
plt.grid(True)

plt.show()
```
